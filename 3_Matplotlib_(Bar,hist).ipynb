# Bar

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x=np.array(["A","B","C","D"])
y=np.array([3,8,1,10])
plt.bar(x,y)
plt.title("Bar plot of X and Y")
plt.show()
```

```python
x=['Apple','Banana']
y=[421,999]
plt.bar(x,y)
plt.show()
```

```python
x=['one','two','three','four','five']
y=[5,20,35,60,15]
plt.title("Vertical Bar plot",color="green",family='monospace',size=20)
plt.xlabel("pen sold",size=15)
plt.ylabel("price",size=15)
plt.bar(x,y)
plt.show()
```

## Horizondal Bar

```python
x=['one','two','three','four','five']
y=[5,20,35,60,15]
plt.barh(x,y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Color

```python
import matplotlib.pyplot as plt
import numpy as np
x=["orange","banana","kiwi"]
y=np.random.randint(100,size=(3))
plt.bar(x,y,color="red")
plt.show()
```

```python
x=["insaf","edwin","shifan","David","vp"]
y=np.random.randint(50,size=(5))
color=['red','yellow','black','pink','purple']
plt.bar(x,y,color=color)
plt.show()
```

## Stacked Bar plot

```python
x=['apple','kiwi','orange']
y1=[5,6,7]
y2=[8,9,10]
plt.bar(x,y1,color='red')
plt.bar(x,y2,color='g',bottom=y1)
plt.show()
```

```python
x=['one','two','three','four']
y1=[2,4,6,8]
y2=[5,6,7,9]
y3=[10,12,15,20]

plt.bar(x,y1,color="r")
plt.bar(x,y2,color='purple',bottom=y1)
plt.bar(x,y2,color='g',bottom=y2)
plt.show()
```

## Line width,hight

```python
x=np.array(["one","two","three","four"])
y=np.array([10,20,30,40])
                 #default is 0.8
plt.bar(x,y,color="black",width=0.2)
plt.show()
```

```python
x=np.array(["one","two","three","four"])
y=np.array([10,20,30,40])
plt.barh(x,y,color="red",height=0.5)
plt.show()
```

## back-to-back Bar

```python
import matplotlib.pyplot as plt
import numpy as np
x=["Apple",'bananaa',"cherry","kiwi"]
y=np.arange(1,5)
plt.bar(x,y,color="r")
plt.bar(x,-y,color='y')
plt.show()
```

```python
x=[10,20,30,40,50]
y1=[2,4,6,8,10]
y2=np.array([3,4,5,6,7])
plt.bar(x,y1,color="red")
plt.bar(x,-y2,color="g")
plt.plot()
```

```python
import numpy as np
import matplotlib.pyplot as plt

barwidth=0.25       #width,hight
plt.figure(figsize=(10,6))

#set height of bar
IT=[12,30,1,8,22]
ECE=[20,6,16,5,10]
CSE=[29,3,24,25,17]


# set position of bar on x axis
br1=np.arange(len(IT))
br2=[x + barwidth for x in br1]
br3=[x+barwidth for x in br2]


plt.bar(br1,IT,color='r',width=barwidth,edgecolor='grey',label='IT')
plt.bar(br2,ECE,color='g',width=barwidth,edgecolor='grey',label='ECE')
plt.bar(br3,CSE,color='b',width=barwidth,edgecolor='grey',label='CSE')



plt.xlabel("Branch",fontweight = 'bold' ,fontsize= 15)
plt.ylabel('students passed',fontweight='bold',fontsize=15)
plt.xticks([r+ barwidth for r in range(len(IT))])
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

N=5
boys=(20,35,30,35,27)
girls=(25,32,34,20,25)
boystd=(2,3,4,1,2)
girlstd=(3,5,2,3,3)
ind=np.arange(N)
width=0.35

fig=plt.subplots(figsize=(10,7))
p1=plt.bar(ind,boys,width,yerr=boystd)
p2=plt.bar(ind,girls,width,bottom=boys,yerr=girlstd)

plt.xlabel("Contribution")
plt.title("contribution by tthe teams")

plt.yticks(np.arange(0,81,10))
plt.legend((p1[0],p2[0]),('boys','girls'))

plt.show()
```
