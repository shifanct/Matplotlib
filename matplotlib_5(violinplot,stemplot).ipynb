# stem plot

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,10,10)
y = np.sin(x)

plt.stem(x,y,linefmt='-.',markerfmt='o',basefmt='k')

plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("stem plot example")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0.1,2*np.pi,41)
y =(np.sin(x))

plt.stem(x,y)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(0.1,2*np.pi,41)
y=np.exp(np.sin(x))

plt.stem(x,y)
plt.show()
```

```python
x = np.linspace(0,2*np.pi,20)
y = np.sin(x)
plt.stem(x,y,basefmt=" ")
plt.title("basic stem plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
x = np.linspace(0,2*np.pi,15)
y = np.cos(x)
plt.stem(x,y,linefmt="r-",markerfmt="bo",basefmt="k",label=("cos(x)"))
plt.title("customized stem plot")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.legend()
plt.show()
```

# fill between|

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.fill_between(x, y1, y2, where=(y1 > y2), color='skyblue', alpha=0.5, label='y1 > y2')
plt.fill_between(x, y1, y2, where=(y1 <= y2), color='coral', alpha=0.5, label='y1 <= y2')

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Fill Between Example')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x=np.linspace(-2,4,100)
y = x**2+2
plt.plot(x,y,label='y=x^2+4x')
plt.fill_between(x,y,color='lightcoral',alpha=0.55)
plt.title("fill between example for quadratic func")
plt.legend()
plt.show()
```
