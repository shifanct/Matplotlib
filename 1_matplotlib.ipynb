# matplotlib labels

```python
xpoints=np.array([80,85,90,95,100,105,110,115,120,125])
ypoints=np.array([240,250,260,270,280,290,290,300,310,320])
plt.plot(xpoints,ypoints,marker="d",linewidth=21,color="pink",ms=30,mec="red",mfc="blue")
plt.xlabel("Average pulse")
plt.ylabel("Calorie Burned")
plt.title("Sports with Data")
plt.show()
```

```python
ypoints=np.array([5,10,15,20])

font1={"family":"serif","color":"blue","size":"20"}
font2={"family":"serif","color":"red","size":"15"}
font3={"family":"monospace","color":"black","size":"25"}
plt.plot(ypoints)
plt.xlabel("Default x points",fontdict=font1,loc="left")  #left,center,right
plt.ylabel('ypoints',fontdict=font2,loc="top")            #bottom,center,top
plt.title("Title for plot",fontdict=font3,loc="right")   #left,right,center
plt.show()
```

# Matplottlib subplots

```python
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(8,10))
xpoints=np.array([2,3,4,5])
ypoints=np.array([5,6,7,8])
plt.subplot(1,2,1)
plt.plot(xpoints,ypoints)

x=[5,6,7]
y=[2,3,4]
plt.subplot(1,2,2)
plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(5,8))
xpoints=np.array([2,3,4,5])
ypoints=np.array([5,6,7,8])
plt.subplot(2,1,1)
plt.plot(xpoints,ypoints)

x=[5,6,7]
y=[2,3,4]
plt.subplot(2,1,2)
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 1)
plt.plot(x,y)

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 3, 2)
plt.plot(x,y)

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 3)
plt.plot(x,y)

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 3, 4)
plt.plot(x,y)

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 5)
plt.plot(x,y)

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 3, 6)
plt.plot(x,y)

plt.show()
```

```python
x=np.array([2,4,8,10,12])
for i in range(1,10):
  plt.subplot(3,3,i)
  plt.plot(x)
plt.show()
```

```python
i=1
x=np.array([2,4,6,5,8,10])
while i<10:
   plt.subplot(3,3,i)
   plt.plot(x,marker="D",linewidth=15,c='hotpink',ms=20,mec="g",mfc="k")
   i=i+1
plt.show()
```

```python
x=[1,2,3,4]
y=[2,3,4,5]

plt.subplot(1,2,1)
plt.plot(x,y)

x=[5,10,15,20]
y=[10,20,30,40]
plt.subplot(1,2,2)
plt.plot(x,y)
plt.subplots_adjust(left=0.10,right=0.5,top=0.9,bottom=0.2,hspace=0.9,wspace=0.5)

plt.show()
```

```python
# Labels of Subplots
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("title 1")

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2,2,2)
plt.plot(x,y)
plt.title("Title 2")

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2,2,3)
plt.plot(x,y)
plt.title("Title 3")

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 2, 4)
plt.plot(x,y)
plt.title("title 4")
plt.suptitle("Supertitle")
plt.show()
```
