 # PIE CHARTS

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([5,25,30,20,20])
plt.pie(x)
plt.show()
```

## startangle

```python
y=np.array([10,20,40,30])
plt.pie(y,startangle=90)
plt.show()
```

## Labels

```python
x=np.array([50,25,25])
mylabel=["fifteen","ten","ten"]
plt.pie(x,startangle=180,labels=mylabel)
plt.show()
```

## explode

```python
import matplotlib.pyplot as plt
import numpy as np
y=np.array([15,25,30,20,20])
mylabel=np.array(["apple","orange","kiwi","pinapple",'grapes'])
myexplode=[0.0,0.0,0.4,0.0,0.0]
plt.pie(y,labels=mylabel,startangle=90,explode=myexplode)
plt.show()
```

## shadow

```python
y=np.array([20,20,20,20,20])
label=["one","two","three","four","five"]
plt.pie(y,startangle=0,labels=label,explode=[0.2,0.0,0.0,0.0,0.2],shadow=True)
plt.show()
```

## color

```python
y=[30,30,30,10]
colr=['red','orange','purple','yellow']
plt.pie(y,labels=["apple",'orange','grapes',"banana"],colors=colr)
plt.show()
```

## Labelspacing

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,5])
y=np.array([3,3,3,3])
plt.plot(x,y,label="line1")
plt.plot(y,x,label="line2")
plt.plot(x,np.sin(x),label="curve3")
plt.plot(x,np.cos(x),label='curve4')
plt.legend(labelspacing=1)
plt.title("multiple line plot")
plt.show()
```

# tight_layout

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,100)
plt.plot(x,np.sin(x),label="sin(x)")
plt.plot(x,np.cos(x),label="cos(x)")
plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

# overlapped stacked graph

```python
import matplotlib.pyplot as plt
import numpy as np

group_a_ages=[22,25,29,30,34,36,28,31,22,25,30,28]
group_b_ages=[24,26,30,32,35,38,27,29,26,30,32,33]

bins=np.arange(20,40,5)


plt.hist(group_a_ages,bins=bins,ec='k',alpha=0.7,label='Group A',stacked=True)
plt.hist(group_b_ages,bins=bins,ec='k',alpha=0.7,label='Group B',stacked=True)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("stacked Histogram of Age Distribution")
plt.legend()


plt.tight_layout()
plt.show()
```

## overlayed histogram with overlapping axis labels

```python
import matplotlib.pyplot as plt
import numpy as np
data1=np.random.normal(50,10,100)
data2=np.random.normal(60,15,100)
bins=np.linspace(20,100,20)

plt.figure(figsize=(6,6))

plt.hist(data1,bins=bins,color="blue",label='Data1')
plt.hist(data2,bins=bins,color='orange',alpha=0.6,label='Data2')

plt.xlabel("very long label for the x axis that might overlap with tick labels")
plt.ylabel("frequancy of occurence might overlap with bars (might overlap with bars)")
plt.title("overlayed histogram with overlapping axis labels (overlapping axis labels)")

plt.grid()
# plt.tight_layout()
plt.legend()
plt.show()
```

# Nested pie chart

```python
outer_labels=["fruits","vegetables"]
outer_sizes=[60,40]

inner_labels=["apples","bananas","carrots","broccoli"]
inner_sizes=[30,30,20,20]

outer_colors=["skyblue","lightgreen"]
inner_colors=["lightcoral","orange","yellowgreen","mediumseagreen"]

#plotting the outer ring
fig,ax=plt.subplots()
ax.pie(outer_sizes,labels=outer_labels,colors=outer_colors,radius=1,wedgeprops=dict(width=0.3,edgecolor="w"))

#plotting the inner ring
ax.pie(inner_sizes,labels=inner_labels,colors=inner_colors,radius=0.7,wedgeprops={"width":0.3,"edgecolor":"w"})

plt.title("Nested pie chart example")
plt.show()
```

# Annotate each data point with its value

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,15,13,17,16]
# plot the x and y axis
plt.plot(x,y,marker="o",color="b",linestyle='solid',label='sample Line')

# Add an annotation for the highest point
plt.annotate("HIGHEST POINT",xy=(4,17),xytext=(3.5,18))

#Annotate each data points with its value
for i,j in zip(x,y):
  plt.text(i,j+0.5,f"({i,j})",ha="center",color="green")


# Labels and title
plt.xlabel("X Axis")
plt.ylabel("Y Label")
plt.title("Line chart with annotation",loc="left")

plt.show()
```
